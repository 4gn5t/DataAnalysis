```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_values_f2, y_values_f2 = [], []
x_values_noise, y_values_noise = [], []

path_funct = './files/Funct_2.txt'
path_noise = './files/Funct_with_GausseNoise.txt'

with open(path_noise, 'r') as file:
    for line in file:
        x, y = map(float, line.split())
        x_values_noise.append(x)
        y_values_noise.append(y)


with open(path_funct, 'r') as file:
    for line in file:
        x, y = map(float, line.split())
        x_values_f2.append(x)
        y_values_f2.append(y)

plt.figure(figsize=(12, 6))
plt.plot(x_values_f2, y_values_f2, label='Funct_2')
plt.plot(x_values_noise, y_values_noise, label='Funct_with_GausseNoise')
plt.legend()
plt.grid()
plt.title('Wave function and GausseNoise')
plt.show()
```

```python
path_update_avg = './files/Moving_update_avg.txt'

x_value_move_avg, y_value_move_avg = [], []

with open(path_update_avg, 'r') as file:
    for line in file:
        x, y = map(float, line.split())
        x_value_move_avg.append(x)
        y_value_move_avg.append(y)

plt.figure(figsize=(12, 6))
plt.plot(x_value_move_avg, y_value_move_avg, label='Moving_average_avg')
plt.plot(x_values_noise, y_values_noise, label='Funct_with_GausseNoise')
plt.legend()
plt.grid()
plt.title('GausseNoise and Method of moving average')
plt.show()
```
